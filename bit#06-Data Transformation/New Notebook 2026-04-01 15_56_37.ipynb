```python
person_list = [(100, "Prashant", 30),
             (101, "David", None),
             (102, "Sushant",None),
             (103, "Abdul", 45),
             (104, "Shruti", 28)]
             
person_df = spark.createDataFrame(person_list).toDF("id", "name", "age")
display(person_df)
```

```python
from pyspark.sql.functions import col
df = (
    person_df.filter(col('age')==28)
)
display(df)
```

```python
from pyspark.sql.functions import col

person_df.filter(col("age").isNull()).display()

#person_df.filter(col("age").isNotNull()).display() (for not Null)
```

```python
person_df.withColumn("age_gt_29", col('age')>29).display()
```

```python
person_df.filter("age > 29").display()
```

How mathametical operators work on null.
Calculate experience for every employee using the following formula.
experience = age - 23


```python
person_df.withColumn('experience',col("age")-23).display()
```

3.4. Calculate the experience knowing that age could be null.
If age is null then assume 23 years for experience calculation.

```python
from pyspark.sql.functions import col, coalesce, lit

person_df.withColumn(
    'experience',
    (coalesce(col("age"), lit(23)) - 23)
).display()
```
