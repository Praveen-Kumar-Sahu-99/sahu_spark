```python
retail_df = (
    spark.read.format('csv')
    .option('header', 'true')
    .option('inferSchema', 'true')
    .load(path="/Volumes/dev/spark_db/datasets/spark_programming/data/invoices.csv")
)
```

```python
retail_df.show()
```

####2. Calculate total_value for each invoice line item

```python
from pyspark.sql.functions import col, round

retail_df.withColumn("total_value", round(col("quantity") * col("unitprice"), 2)).display()
```

#### using experssion as variable 

```python
total_value_expr_1 = round(col("quantity") * col("unitprice"), 2)

retail_df.withColumn("total_value", total_value_expr_1).display()
```
